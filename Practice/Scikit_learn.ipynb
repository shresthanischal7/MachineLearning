```python
import pandas as pd
import numpy as np
import sklearn.preprocessing as sk
df = pd.read_csv('obesity_prediction.csv')
```

```python
df.head()
```

```python
y = df['Obesity']
label_enc = sk.LabelEncoder()
y_enc = label_enc.fit_transform(y)
np.unique(y_enc)
```

```python
df['Obesity'].unique()
```

```python
one_enc = sk.OneHotEncoder()
enc_df = pd.DataFrame(one_enc.fit_transform(df[['Obesity']]).toarray())
enc_df
```

```python
one_enc.inverse_transform(enc_df)
```
